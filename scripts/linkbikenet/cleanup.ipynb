Temporary script to clean up files produced by lbn v0.7.2

```python
import os
import geopandas as gpd
```

```python
path_in = '/Users/mszell/Tresorit/bikenetkitshare/latest/linkbikenet_old/'
path_out = '/Users/mszell/Tresorit/bikenetkitshare/latest/linkbikenet/'
keepedgedata = ['length', 'lcc_step', 'geometry']
```

```python
for f in os.listdir(path_in):
    if not f.startswith("."):
        # Existing bike networks, just take one of them
        if f.endswith('closest_components-existing_bike_network.geojson'):
            geoj = gpd.read_file(path_in+f)
            for p in geoj.keys():
                if p not in keepedgedata:
                    del geoj[p] 
            geoj.to_file(path_out+f.split("-")[0]+"-existing_bike_network.geojson", driver="GeoJSON", RFC7946="YES")
        # Other files just copy over
        elif not f.endswith('existing_bike_network.geojson'):
            geoj = gpd.read_file(path_in+f)
            geoj.to_file(path_out+f, driver="GeoJSON", RFC7946="YES")
            pass
```
