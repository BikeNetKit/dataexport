Temporary script to change and resave original files with new CRS and RFC7946.

```python
import os
import geopandas as gpd
```

```python
path_in = '../../dataexports/latest/sortbikenet_old/'
path_out = '../../dataexports/latest/sortbikenet/'
```

```python
for f in os.listdir(path_in):
    if f.endswith('geojson'):
        geoj = gpd.read_file(path_in+f)
        geoj.rename(columns={"order": "ordering"}, inplace=True)
        if "ordering" in geoj:
            if "from_scratch" in f:
                geoj["ordering"] = geoj["ordering"] + 1
            else:
                geoj.drop(geoj[geoj["ordering"] == 0].index, inplace=True) # Drop existing bike networks
        geoj.to_crs(epsg='4326', inplace=True)
        geoj.to_file(path_out+f, driver="GeoJSON", RFC7946="YES")
```
