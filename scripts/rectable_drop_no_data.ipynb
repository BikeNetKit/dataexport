Auxiliary script to update the rectable after the fact, to set all to zero where data is missing (could not be exported due to bugs or other issues).

```python
import pandas as pd
import os
```

```python
cities_path = '/Users/mszell/Tresorit/bikenetkitshare/'
toolnames = {
    'growbikenet': ['gbn','gbn+'], 
    'fixbikenet': ['fbn'],
    'linkbikenet': ['lbn'],
    'superblockify': ['sb'],
}
```

## Read rectable and set zeros

```python
rectable = pd.read_csv(cities_path+'latest/meta/rec_table.csv')
```

```python
for tool,cols in toolnames.items():
    for i,r in rectable.iterrows():
        if not any(f.startswith(r.cityid) for f in os.listdir(cities_path+'latest/'+tool)):
            print(tool+" has no export data for "+r.cityid)
            for col in cols:
                rectable.at[i, col] = 0
    print()
```

## Overwrite rectable

```python
rectable.to_csv('../cities/meta/rec_table.csv', index=False)
rectable.to_csv(cities_path+'latest/meta/rec_table.csv', index=False)
```
