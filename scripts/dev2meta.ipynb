Auxiliary script to clean and copy the development city file to the meta folder.

```python
import csv
from growbikenet.functions import *
```

```python
cities_path = "../cities/"
cityfilename_in = "european_capitalsand100000pop.csv"
cityfilename_out = "cities.csv"
```

## Read data from development

```python
with open(cities_path+'development/'+cityfilename_in, mode='r') as infile:
    reader = csv.reader(infile, delimiter=";")
    header = ['cityid'] + next(reader)
    cities = [{
                header[0]: slugify(rows[0])+"_"+slugify(rows[3]),
                header[1]: rows[0], 
                header[2]: rows[1], 
                header[3]: rows[2], 
                header[4]: rows[3], 
                header[5]: str(int(float(rows[4]))),
              } 
              for rows in reader]
```

## Clean and write data to meta

```python
with open(cities_path+'meta/'+cityfilename_out, 'w', newline='') as outfile:
    writer = csv.DictWriter(outfile, fieldnames=header, delimiter=";")
    writer.writeheader()
    writer.writerows(cities)
```
