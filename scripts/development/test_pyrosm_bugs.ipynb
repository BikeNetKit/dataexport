# Service road issue

## OSMnx

```python
import osmnx as ox
g_ox = ox.graph.graph_from_place("Lyon", network_type="drive")
ec = ["r" if data["highway"] == "service" else "w" for u, v, key, data in g_ox.edges(keys=True, data=True)]
ox.plot_graph(g_ox, node_size=0, edge_color=ec);
```

## Pyrosm

```python
import pyrosm
fp = pyrosm.get_data("Lyon")
osm = pyrosm.OSM(fp)
(street_nodes, street_edges) = osm.get_network(network_type="driving", nodes=True)
g_pyrosm = pyrosm.OSM.to_graph(street_nodes, street_edges, graph_type='networkx', osmnx_compatible=True, simplify=True)
ec = ["r" if data["highway"] == "service" else "w" for u, v, key, data in g_pyrosm.edges(keys=True, data=True)]
ox.plot_graph(g_pyrosm, node_size=0, edge_color=ec);
```
