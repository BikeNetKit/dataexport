Order cities by population.

```python
import pandas as pd
```

```python
cities_path = "../cities/meta/"
cityfilename = "cities.csv"
cities = pd.read_csv(cities_path+cityfilename, sep=";")
cities.sort_values('population', inplace=True)
```

```python
cities.to_csv(cities_path+"cities_by_population.csv", sep=';', index=False)
```
