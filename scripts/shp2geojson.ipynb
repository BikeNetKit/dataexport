Auxiliary script to turn a shp file into geojson. Also, turn MultiPoints into Points.

```python
import os
import geopandas as gpd
```

```python
inpath = '/Users/mszell/Github/BikeNetKit/FixBikeNet/tests/test_data/'
outpath = '/Users/mszell/Github/BikeNetKit/FixBikeNet/tests/test_data/'
```

```python
for f in os.listdir(inpath):
    if not f.startswith(".") and f.endswith(".shp"):
        gdf = gpd.read_file(inpath+f)
        split_tup = os.path.splitext(f)
        file_name = split_tup[0]
        file_extension = split_tup[1]

        # MultiPoints to Points
        # https://github.com/SiRan-Chen/Converting-Multipoint-Data-to-Single-Point-Data/blob/main/dandian.py
        single_points = []
        for idx, row in gdf.iterrows():
            geom = row.geometry
            if geom.geom_type == "MultiPoint":
                for point in geom.geoms:
                    new_row = row.copy()
                    new_row.geometry = point
                    single_points.append(new_row)
            else:
                single_points.append(row)
        single_gdf = gpd.GeoDataFrame(single_points, columns=gdf.columns, crs=gdf.crs)

        single_gdf.to_file(outpath+file_name+".geojson", driver="GeoJSON", RFC7946="YES")
```
