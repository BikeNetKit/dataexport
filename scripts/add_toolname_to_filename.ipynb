Add toolname in front of filenames (after cityid) where missing. In place!

```python
import os
```

```python
toolname = "growbikenet"
path = '/Users/mszell/Tresorit/bikenetkitshare/latest/test/'
```

```python
for f in os.listdir(path):
    if not f.startswith("."):
        if toolname not in f:
            split_tup = os.path.splitext(f)
            file_name = split_tup[0]
            file_extension = split_tup[1]
            fs = file_name.split("-")
            os.rename(path+f, path+fs[0]+"-"+toolname+"-"+fs[1]+file_extension)
```
